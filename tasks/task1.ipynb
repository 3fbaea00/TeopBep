### Задача 1.8
В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

а)

$$n_a = 10^5$$

```python
n_a = 10 ** 5
print(f"а) {n_a}")
```

б)

$$n_b = C_{10+5-1}^{5} = C_{14}^{5}$$

```python
n_b = sc.comb(10 + 5 - 1, 5)
print(int(n_b))
```

### Задача 1.9
Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 – по 2 раза?

$$n = \frac{7!}{3! \cdot 2! \cdot 2!}$$

```python
n_1_9 = sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2))
print(f"{int(n_1_9)}")
```

### Задача 1.10
Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

Событие A - получилось слово ТОР

$$P(A) = \frac{1}{A_6^3} = \frac{1}{120}$$

```python
total_a = sc.perm(6, 3)
prob_a = 1 / total_a
print(f"а) ТОР: {round(prob_a, 6)}")
```

Событие B - получилось слово ТЕОРИЯ

$$P(B) = \frac{1}{6!} = \frac{1}{720}$$

```python
total_b = sc.factorial(6)
prob_b = 1 / total_b
print(f"б) ТЕОРИЯ: {round(prob_b, 6)}")
```

### Задача 1.11
Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

$$m = \frac{6!}{3! \cdot 2! \cdot 1!} = 60, \quad P = \frac{1}{60}$$

```python
m_1_11 = sc.factorial(6) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(1))
prob_1_11 = 1 / m_1_11
print(f" АНАНАС: {round(prob_1_11, 6)}")
```

### Задача 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

Событие A – чередование

$$P(A) = \frac{2 \cdot 5! \cdot 5!}{10!}$$

```python
prob_16a = (2 * sc.factorial(5) * sc.factorial(5)) / sc.factorial(10)
print(f"а) {round(prob_16a, 6)}")
```

Событие B – все супруги сидят рядом

$$P(B) = \frac{4! \cdot 2^5}{9!}$$

```python
prob_16b = (sc.factorial(4) * (2 ** 5)) / sc.factorial(9)
print(f"б) {round(prob_16b, 6)}")
```

### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

Общее число исходов: $$m = A_{36}^{7}$$

```python
total_17 = sc.perm(36, 7)
```

Событие A - в два купе

$$P(A) = \frac{C_9^2 \cdot A_8^7}{A_{36}^{7}}$$

```python
fav_a = sc.comb(9, 2) * sc.perm(8, 7)
prob_17a = fav_a / total_17
print(f"а) {round(prob_17a, 8)}")
```

Событие B - в семь купе

$$P(B) = \frac{C_9^7 \cdot 4^7 \cdot 7!}{A_{36}^{7}}$$

```python
fav_b = sc.comb(9, 7) * (4 ** 7) * sc.factorial(7)
prob_17b = fav_b / total_17
print(f"б) {round(prob_17b, 8)}")
```

Событие C - в три купе (все три заняты

$$P(C) = \frac{C_9^3 \cdot \left( A_{12}^7 - 3 \cdot A_8^7 \right)}{A_{36}^{7}}$$

```python
fav_c = sc.comb(9, 3) * (sc.perm(12, 7) - 3 * sc.perm(8, 7))
prob_17c = fav_c / total_17
print(f"в) {round(prob_17c, 8)}")
```
