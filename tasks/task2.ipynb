### Задача 1.38
Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

### Задача 1.43
Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.

Общее число номеров: $9 \cdot 10^4 = 90000$.

а) все цифры различные: первая цифра – 9 вариантов, остальные 4 выбираются упорядоченно из 9 оставшихся: $9 \cdot P(9,4) = 9 \cdot 9 \cdot 8 \cdot 7 \cdot 6 = 27216$. Вероятность $= \frac{27216}{90000}$.

б) все цифры одинаковые: возможны только цифры 1–9, т.е. 9 номеров. Вероятность $= \frac{9}{90000}$.

в) все цифры нечетные: нечетных цифр 5 (1,3,5,7,9), каждая из 5 позиций может быть любой из них, всего $5^5 = 3125$. Вероятность $= \frac{3125}{90000}$.

```python
total = 9 * 10**4

fav_a = 9 * sc.perm(9, 4)
prob_a = fav_a / total

prob_b = 9 / total

prob_c = 5**5 / total

print(round(prob_a, 4), round(prob_b, 4), round(prob_c, 4))
```

### Задача 1.48
В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

Всего исходов: $6^3 = 216$.

а) число способов получить сумму 11: $x_1+x_2+x_3=11$, $1 \le x_i \le 6$. Замена $y_i = x_i - 1$: $y_1+y_2+y_3=8$, $0 \le y_i \le 5$. Без ограничений $\binom{10}{2}=45$, вычитаем случаи $y_i \ge 6$: $3 \cdot \binom{4}{2}=18$, получаем $45-18=27$. Вероятность $= \frac{27}{216}$.

б) выигрыш – сумма больше 10, т.е. 11,12,…,18. Число благоприятных исходов равно 108 (по симметрии распределения или прямым подсчётом). Вероятность $= \frac{108}{216} = 0.5$.

```python
total_dice = 6**3

ways_11 = 27
prob_11 = ways_11 / total_dice

ways_win = 108
prob_win = ways_win / total_dice

print(round(prob_11, 4), round(prob_win, 4))
```
